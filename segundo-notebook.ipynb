```python
#importadond o seaborn para plotar os dados
import seaborn as sns
```

```python
grafico = sns.histplot(data=df,x="categoria",kde=True)
grafico.set_title("categorias mais vendidas")
grafico.set_xlabel("categorias")
grafico.set_ylabel("vendas")
```

```python
grafico = sns.boxplot(data=df,x="categoria",y="preco_unitario")
grafico.set_title("Preco por categoria")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Precos")
```
